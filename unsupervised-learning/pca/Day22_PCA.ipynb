# Chapter 22 — Principal Component Analysis (PCA)

PCA is a dimensionality-reduction technique that transforms correlated or high-dimensional features into a smaller set of principal components while preserving as much variance as possible.

**Dataset:** Iris dataset  
**Goal:** Reduce 4 features to 2 principal components and visualize the result.

## 1. Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

## 2. Load the Iris Dataset

```python
iris = load_iris()
X = iris.data
y = iris.target

df = pd.DataFrame(X, columns=iris.feature_names)
df.head()
```

## 3. Check Dataset Shape

```python
print("Dataset shape:", X.shape)
print("Number of features:", X.shape[1])
```

The Iris dataset contains **150 samples and 4 features**.

## 4. Standardize the Features

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(X_scaled[:5])
```

Standardization puts the features on a comparable scale before PCA is applied.

## 5. Apply PCA — 4D to 2D

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Original shape:", X_scaled.shape)
print("PCA shape:", X_pca.shape)
```

The representation changes from **4 original features to 2 principal components**.

## 6. Explained Variance

```python
print("Explained variance ratio:")
print(pca.explained_variance_ratio_)

total_variance = pca.explained_variance_ratio_.sum()
print("Total variance retained:", total_variance)
print("Percentage retained:", total_variance * 100)
```

The explained variance ratio tells us how much of the dataset's total variance is captured by each principal component. For standardized Iris data, the first two components retain approximately **96%** of the variance.

## 7. Create a PCA DataFrame

```python
pca_df = pd.DataFrame(X_pca, columns=["PC1", "PC2"])
pca_df["Target"] = y
pca_df.head()
```

## 8. Visualize the PCA Components

```python
plt.figure(figsize=(8, 6))
plt.scatter(pca_df["PC1"], pca_df["PC2"], c=pca_df["Target"], s=50)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA — Iris Dataset")
plt.show()
```

The 2D projection makes the structure of the Iris dataset easier to visualize.

## 9. Variance Captured by Each Component

```python
for i, variance in enumerate(pca.explained_variance_ratio_, start=1):
    print(f"PC{i}: {variance * 100:.2f}%")
```

## 10. Fit PCA with All Components

```python
pca_full = PCA()
X_pca_full = pca_full.fit_transform(X_scaled)

explained_variance = pca_full.explained_variance_ratio_
```

## 11. Scree Plot

```python
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(explained_variance) + 1), explained_variance, marker="o")
plt.xlabel("Principal Component")
plt.ylabel("Explained Variance Ratio")
plt.title("PCA Explained Variance")
plt.xticks(range(1, 5))
plt.grid(True)
plt.show()
```

## 12. Cumulative Explained Variance

```python
cumulative_variance = np.cumsum(pca_full.explained_variance_ratio_)

for i, variance in enumerate(cumulative_variance, start=1):
    print(f"{i} components: {variance * 100:.2f}%")

plt.figure(figsize=(8, 5))
plt.plot(range(1, len(cumulative_variance) + 1), cumulative_variance, marker="o")
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Cumulative Explained Variance")
plt.xticks(range(1, 5))
plt.grid(True)
plt.show()
```

## Key Takeaways

- PCA reduces the number of dimensions while preserving important variance.
- Principal components are new combinations of the original features.
- Standardization is important when features have different scales.
- Explained variance helps decide how many components to retain.
- In this Iris example, 2 components preserve approximately 96% of the variance.

### Pipeline
`Iris Dataset → Standardize → PCA → 2D Representation → Visualization`
